```python
%sql
CREATE OR REPLACE TABLE workspace.gold.customer_summary AS
SELECT
  custkey,
  COUNT(*) AS total_events,
  SUM(CASE WHEN event_type = 'page_view' THEN 1 ELSE 0 END) AS page_views,
  SUM(CASE WHEN event_type = 'purchase' THEN 1 ELSE 0 END) AS purchases,
  SUM(CASE WHEN event_type = 'refund' THEN 1 ELSE 0 END) AS refunds,
  SUM(amount) AS net_web_revenue,
  MAX(order_count) AS lifetime_orders,
  MAX(total_order_value) AS lifetime_order_value,
  MAX(event_time) AS last_activity
FROM workspace.silver.customer_events
GROUP BY custkey;
```

```python
%sql
SELECT * FROM workspace.gold.customer_summary ORDER BY custkey;
```
